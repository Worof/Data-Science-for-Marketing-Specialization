```python
%%R

# Find regression coefficients for SLR model predicting Sales from Advertising;

# Obtain estimates of slope and intercept by creating an SLR model named “fit” and displaying it;

fit <- lm(Sales ~ Advertising);

fit;
```

```python
%%R

# Summarize the details of the SLR model;

summary(fit);
```

```python
%%R

# Visualize the residuals for the SLR model;

# Create a quick visualization of the residuals distribution;

plot(density(resid(fit)));
```

```python
%%R

# Create a better-looking visualization;

plot(density(resid(fit)) , cex = 1.1, cex.lab = 2, cex.axis = 2,cex.main = 2, main = "Kernel density estimate of residuals distribution",xlab = "value",ylab = "density");
```

```python
%%R

# Create a QQ plot to check visually for normality of residuals;

# Plot observed quantiles of the ECDF against quantiles predicted for a normal distribution;

qqnorm(resid(fit), cex = 1.1, cex.lab = 1.5, cex.axis = 1.5,cex.main = 4);
```

```python
%%R

# Create a QQ plot to check visually for normality of residuals;
# Plot observed quantiles of the ECDF against quantiles predicted for a normal distribution;
qqnorm(resid(fit), cex = 1.1, cex.lab = 1.5, cex.axis = 1.5,cex.main = 4);

# Plot comparison line;
qqline(resid(fit));
```

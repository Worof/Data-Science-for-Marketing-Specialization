```python
%%R

# View numerical values of mean and variance of Sales and Price variables in Carseats dataset;

mean(Sales); var(Sales); mean(Price); var(Price);
```

```python
%%R

# Create a scatter plot with a red horizontal line through the mean value for Sales and red dashed lines at the mean plus and minus 2*sd(Sales);

plot(Price, Sales, cex = 1.6, cex.lab = 1.6, cex.axis = 1.6, pch = 16);

abline(h = mean(Sales), lwd = 3, col = "red");

abline(h = mean(Sales) + 2*sd(Sales), lwd = 3, col = "red", lty = 3);

abline(h = mean(Sales) - 2*sd(Sales), lwd = 3, col = "red", lty = 3);

ecdf(Sales)(mean(Sales) + 2*sd(Sales)) -ecdf(Sales)(mean(Sales) - 2*sd(Sales));

ecdf(Price)(mean(Price) + 2*sd(Price)) -ecdf(Price)(mean(Price) - 2*sd(Price));
```

```python
%%R

# View numerical values of mean and standard deviation of Sales;

mean(Sales); sd(Sales);
```

```python
%%R

# Find approximate 95% confidence interval for the mean of Sales;

mean(Sales) + 1.96*sd(Sales);

mean(Sales) - 1.96*sd(Sales);
```

```python
%%R

# Create a scatter plot with a regression line for predicting Sales from Price;

# Plot scatter plot for Sales vs.Price;

plot(Price, Sales, cex = 1.6, cex.lab = 1.6, cex.axis = 1.6, pch = 16);
```

```python
%%R

# Create a scatter plot for Sales vs.Price and add a least-squares regression line in blue;

plot(Price, Sales, cex = 1.6, cex.lab = 1.6, cex.axis = 1.6, pch = 16);
abline(lm(Sales ~ Price), col = "blue", cex = 1.4, cex.lab = 1.6, cex.axis = 1.6, pch = 16,lwd = 3);
```

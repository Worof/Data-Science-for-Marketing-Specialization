```python
%%R
# Create data frame consisting of just Price and Sales;

df <- cbind(Price, Sales)
```

```python
%%R
# Estimate the optimal number of clusters;

fviz_nbclust(df, kmeans, method = "silhouette");

fviz_nbclust(df, kmeans, method = "wss") + geom_vline(xintercept = 3, linetype = 2);
```

```python
%%R
# Show table of 3 k-means clusters;

kmeans(df, centers = 3, iter.max = 10, nstart = 3)
```

```python
%%R
# Show table of 3 k-means clusters;

kmeans(df, centers = 3, iter.max = 10, nstart = 3)
```

```python
%%R
# Store the k-means results in a variable called 'km.res';
km.res <- kmeans(df, centers = 3, iter.max = 10, nstart = 3)

# Create a better-looking plot;
fviz_cluster(km.res, df, ellipse.type = "convex")+ theme(axis.title.y = element_text(size = rel(2), angle = 90)) + theme(axis.title.x = element_text(size = rel(2), angle = 0))+ theme(axis.text.x = element_text(angle = 0, hjust = 1, size=18,color="black")) +  theme(axis.text.y = element_text(angle = 0, hjust = 1, size=18,color="black"));
```

```python
%%R

# Show mean and standard deviation of Price;

mean(Price); sd(Price);
```

```python
%%R
# Fit a smooth curve, colored red, to the scatter plot;

if (!require(car)) install.packages('car'); library(car);

scatter.smooth(Price, Sales, cex = 1.6, cex.lab = 1.6, cex.axis = 1.6, pch = 16, lpars = list(col = "red", lwd = 3));
```

```python
%%R
# Add a blue regression line to the scatter plot;

plot(Price, Sales, cex = 1.6, cex.lab = 1.6, cex.axis = 1.6, pch = 16);

abline(lm(Sales ~ Price, data = Carseats), col = "blue", cex = 1.4, cex.lab = 1.6, cex.axis = 1.6, pch = 16,lwd = 3);
```
